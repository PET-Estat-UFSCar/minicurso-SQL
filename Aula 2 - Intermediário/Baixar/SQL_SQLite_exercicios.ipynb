Este notebook usa a mesma base da apresentação: clientes, pedidos, produtos e itens_pedido. A base é criada pelo arquivo utils/gerar_dados.py, com semente fixa, para que os resultados sejam reproduzíveis.

Os exercícios praticam inspeção de schema, PK/FK, relações 1:N e N:N, INNER JOIN, LEFT JOIN, agregações e identificação de registros sem correspondência.

No Google Colab, a primeira célula baixa o gerador oficial do repositório e cria minicurso_aula2.db na sessão. Não é necessário usar a base da livraria.

```python
import sqlite3
from pathlib import Path
from urllib.request import urlretrieve

ARQUIVO_GERADOR = Path("gerar_dados.py")
ARQUIVO_BANCO = Path("minicurso_aula2.db")
URL_GERADOR = (
    "https://raw.githubusercontent.com/PET-Estat-UFSCar/minicurso-SQL/"
    "main/Aula%202%20-%20Intermedi%C3%A1rio/utils/gerar_dados.py"
)

if not ARQUIVO_GERADOR.exists():
    urlretrieve(URL_GERADOR, ARQUIVO_GERADOR)

namespace = {}
exec(ARQUIVO_GERADOR.read_text(encoding="utf-8"), namespace)
namespace["criar_banco"](ARQUIVO_BANCO)

conexao = sqlite3.connect(ARQUIVO_BANCO)
conexao.execute("PRAGMA foreign_keys = ON")
cursor = conexao.cursor()
print(f"Banco conectado: {ARQUIVO_BANCO.resolve()}")
```
